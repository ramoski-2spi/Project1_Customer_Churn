```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (classification_report, confusion_matrix, accuracy_score, roc_auc_score, RocCurveDisplay)
```

```python
#import the training and test data
X_train = pd.read_csv('../data/X_train.csv')
X_test = pd.read_csv('../data/X_test.csv')
y_train = pd.read_csv('../data/y_train.csv').squeeze()
y_test = pd.read_csv('../data/y_test.csv').squeeze()
```

```python
#Initialize logistic regression model
model = LogisticRegression(class_weight= 'balanced', max_iter= 5000)
```

```python
#train the model
model.fit(X_train, y_train)
```

```python
#make predictions
predictions = model.predict(X_test)
probabilities = model.predict_proba(X_test)[:, 1]
```

```python
#Accuracy evaluation
accuracy_score(y_test, predictions)
```

```python
#Confusion matrix
print(confusion_matrix(y_test, predictions))
```

```python
#Classification report
print(classification_report(y_test, predictions))
```

```python
#ROC AUC to see how well the base model differentiate positive and negatif cases.
roc_auc_score(y_test, probabilities)
```

```python
RocCurveDisplay.from_predictions(y_test, probabilities)
```

What we did: 

A baseline Logistic Regression model was trained to predict customer churn. Model performance was evaluated using accuracy, precision, recall, F1-score, and ROC-AUC. These results provide a reference point for future model improvements.
